```python
# CSF Data Analyst Intern Assessment

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
import glob
import re

%matplotlib inline

# -----------------------------
# 1. Load files
# -----------------------------

cheese_file = glob.glob("cheese_data*.csv")[0]
temp_file = glob.glob("Canada_Temperature_Data*.csv")[0]
weather_file = glob.glob("canada_weather*.csv")[0]

cheese = pd.read_csv(cheese_file)
temp = pd.read_csv(temp_file)
weather = pd.read_csv(weather_file)

print("Loaded files:")
print(cheese_file)
print(temp_file)
print(weather_file)

display(cheese.head())
display(temp.head())
display(weather.head())

# -----------------------------
# 2. Clean data
# -----------------------------

cheese.columns = cheese.columns.str.strip()
temp.columns = temp.columns.str.strip()
weather.columns = weather.columns.str.strip()

cheese = cheese.rename(columns={"ManufacturerProvCode": "Province"})
temp = temp.rename(columns={"Prov": "Province"})

cheese["Province"] = cheese["Province"].astype(str).str.strip().str.upper()
temp["Province"] = temp["Province"].astype(str).str.strip().str.upper()

weather["Province"] = (
    weather["Community"]
    .astype(str)
    .str.split(",")
    .str[-1]
    .str.strip()
    .str.upper()
)

cheese = cheese.drop_duplicates()
temp = temp.drop_duplicates()
weather = weather.drop_duplicates()

cheese["MoisturePercent"] = pd.to_numeric(
    cheese["MoisturePercent"], errors="coerce"
)

temp["Tm"] = pd.to_numeric(temp["Tm"], errors="coerce")
temp["Year"] = pd.to_numeric(temp["Year"], errors="coerce")

cheese = cheese.dropna(subset=["Province", "MoisturePercent", "CategoryTypeEn"])
temp = temp.dropna(subset=["Province", "Tm", "Year"])

# -----------------------------
# 3. Clean weather temperature
# -----------------------------

def extract_celsius(value):
    value = str(value).replace("−", "-")
    match = re.search(r"-?\d+\.?\d*", value)
    return float(match.group()) if match else np.nan

weather["AnnualHighC"] = weather["Annual(Avg. high °C (°F))"].apply(extract_celsius)
weather["AnnualLowC"] = weather["Annual(Avg. low °C (°F))"].apply(extract_celsius)

weather["WeatherAvgTemp"] = (
    weather["AnnualHighC"] + weather["AnnualLowC"]
) / 2

weather = weather.dropna(subset=["Province", "WeatherAvgTemp"])

# -----------------------------
# 4. Feature engineering
# -----------------------------

hard_categories = [
    "Firm Cheese",
    "Semi-soft Cheese",
    "Hard Cheese",
    "Semi-hard Cheese"
]

cheese["HardCheese"] = cheese["CategoryTypeEn"].isin(hard_categories)

if "Organic" in cheese.columns:
    cheese["Organic"] = pd.to_numeric(cheese["Organic"], errors="coerce").fillna(0)
else:
    cheese["Organic"] = 0

# -----------------------------
# 5. Aggregation
# -----------------------------

province_cheese = (
    cheese.groupby("Province", as_index=False)
    .agg(
        CheeseCount=("Province", "count"),
        AvgMoisture=("MoisturePercent", "mean"),
        HardCheeseRatio=("HardCheese", "mean"),
        OrganicRate=("Organic", "mean")
    )
)

province_cheese["HardCheesePercent"] = province_cheese["HardCheeseRatio"] * 100
province_cheese["OrganicPercent"] = province_cheese["OrganicRate"] * 100

latest_year = int(temp["Year"].max())
recent_temp = temp[temp["Year"] >= latest_year - 20]

province_temp = (
    recent_temp.groupby("Province", as_index=False)
    .agg(AvgTemperature=("Tm", "mean"))
)

province_weather = (
    weather.groupby("Province", as_index=False)
    .agg(
        WeatherAvgTemp=("WeatherAvgTemp", "mean"),
        WeatherStations=("Community", "count")
    )
)

merged = province_cheese.merge(province_temp, on="Province", how="inner")
merged = merged.merge(province_weather, on="Province", how="left")

print("Final Merged Dataset")
display(merged)

# -----------------------------
# 6. Chart style
# -----------------------------

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.edgecolor": "#DDDDDD",
    "axes.labelcolor": "#333333",
    "xtick.color": "#333333",
    "ytick.color": "#333333",
    "font.size": 11,
    "axes.titleweight": "bold",
    "axes.titlesize": 16
})

def clean_axes(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.grid(axis="y", linestyle="--", alpha=0.25)
    ax.set_axisbelow(True)

# ============================================================
# Visualization 1
# Cheese count by province
# ============================================================

plot_data = merged.sort_values("CheeseCount", ascending=True)

fig, ax = plt.subplots(figsize=(11, 6))

colors = plt.cm.YlOrBr(np.linspace(0.35, 0.9, len(plot_data)))

bars = ax.barh(
    plot_data["Province"],
    plot_data["CheeseCount"],
    color=colors,
    edgecolor="#5C4033"
)

ax.set_title("Cheese Products Listed by Province", pad=15)
ax.set_xlabel("Number of Cheese Products")
ax.set_ylabel("Province")

for bar in bars:
    width = bar.get_width()
    ax.text(
        width + 5,
        bar.get_y() + bar.get_height() / 2,
        f"{int(width)}",
        va="center",
        fontweight="bold"
    )

clean_axes(ax)
ax.grid(axis="x", linestyle="--", alpha=0.25)
plt.tight_layout()
plt.show()

# ============================================================
# Visualization 2
# Temperature vs moisture bubble chart
# ============================================================

bubble_data = merged.copy()

bubble_data["BubbleSize"] = (
    np.sqrt(bubble_data["CheeseCount"]) * 120
)

fig, ax = plt.subplots(figsize=(11, 6.5))

scatter = ax.scatter(
    bubble_data["AvgTemperature"],
    bubble_data["AvgMoisture"],
    s=bubble_data["BubbleSize"],
    c=bubble_data["HardCheesePercent"],
    cmap="viridis",
    alpha=0.75,
    edgecolor="black",
    linewidth=0.8
)

for _, row in bubble_data.iterrows():
    ax.annotate(
        row["Province"],
        (row["AvgTemperature"], row["AvgMoisture"]),
        textcoords="offset points",
        xytext=(6, 5),
        fontsize=10,
        fontweight="bold"
    )

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Firm / Semi-Soft Cheese Share (%)")

ax.set_title("Temperature vs Average Cheese Moisture", pad=15)
ax.set_xlabel("Average Provincial Temperature (°C)")
ax.set_ylabel("Average Cheese Moisture (%)")

clean_axes(ax)
plt.tight_layout()
plt.show()

# ============================================================
# Visualization 3
# Cheese category distribution - percentage stacked bar
# ============================================================

category_table = pd.crosstab(
    cheese["Province"],
    cheese["CategoryTypeEn"]
)

category_table = category_table.loc[
    category_table.sum(axis=1).sort_values(ascending=False).index
]

category_percent = category_table.div(
    category_table.sum(axis=1),
    axis=0
) * 100

print("Cheese Category Count DataFrame")
display(category_table)

print("Cheese Category Percentage DataFrame")
display(category_percent.round(1))

fig, ax = plt.subplots(figsize=(12, 6.5))

category_percent.plot(
    kind="bar",
    stacked=True,
    ax=ax,
    colormap="tab20",
    edgecolor="white",
    linewidth=0.4
)

ax.set_title("Cheese Category Mix by Province (%)", pad=15)
ax.set_xlabel("Province")
ax.set_ylabel("Percentage of Cheese Products")

ax.legend(
    title="Cheese Category",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

clean_axes(ax)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# Visualization 4
# Temperature source validation
# ============================================================

weather_validation = merged.dropna(subset=["WeatherAvgTemp"]).copy()

print("Temperature Validation DataFrame")
display(
    weather_validation[
        ["Province", "AvgTemperature", "WeatherAvgTemp", "WeatherStations", "CheeseCount"]
    ]
)

fig, ax = plt.subplots(figsize=(10, 6))

scatter = ax.scatter(
    weather_validation["AvgTemperature"],
    weather_validation["WeatherAvgTemp"],
    s=np.sqrt(weather_validation["WeatherStations"]) * 180,
    c=weather_validation["CheeseCount"],
    cmap="plasma",
    alpha=0.75,
    edgecolor="black",
    linewidth=0.8
)

for _, row in weather_validation.iterrows():
    ax.annotate(
        row["Province"],
        (row["AvgTemperature"], row["WeatherAvgTemp"]),
        textcoords="offset points",
        xytext=(6, 5),
        fontsize=10,
        fontweight="bold"
    )

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Cheese Count")

ax.set_title("Comparison of Two Temperature Sources", pad=15)
ax.set_xlabel("Historical Temperature Dataset Avg. Temp (°C)")
ax.set_ylabel("Weather Dataset Avg. Temp (°C)")

clean_axes(ax)
plt.tight_layout()
plt.show()

# ============================================================
# Visualization 5
# Correlation heatmap
# ============================================================

corr_data = merged[
    [
        "CheeseCount",
        "AvgMoisture",
        "HardCheesePercent",
        "OrganicPercent",
        "AvgTemperature",
        "WeatherAvgTemp"
    ]
].corr(numeric_only=True)

print("Correlation DataFrame")
display(corr_data.round(2))

fig, ax = plt.subplots(figsize=(8, 6))

heatmap = ax.imshow(
    corr_data,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.colorbar(heatmap, ax=ax, label="Correlation")

ax.set_xticks(np.arange(len(corr_data.columns)))
ax.set_yticks(np.arange(len(corr_data.index)))

ax.set_xticklabels(corr_data.columns, rotation=35, ha="right")
ax.set_yticklabels(corr_data.index)

for i in range(len(corr_data.index)):
    for j in range(len(corr_data.columns)):
        value = corr_data.iloc[i, j]
        ax.text(
            j,
            i,
            f"{value:.2f}",
            ha="center",
            va="center",
            color="black",
            fontweight="bold"
        )

ax.set_title("Correlation Matrix: Climate and Cheese Variables", pad=15)

plt.tight_layout()
plt.show()

# ============================================================
# Visualization 6
# Hard cheese percentage by province
# ============================================================

hard_plot = merged.sort_values("HardCheesePercent", ascending=True)

fig, ax = plt.subplots(figsize=(11, 6))

colors = plt.cm.GnBu(np.linspace(0.35, 0.9, len(hard_plot)))

bars = ax.barh(
    hard_plot["Province"],
    hard_plot["HardCheesePercent"],
    color=colors,
    edgecolor="#333333"
)

ax.set_title("Firm / Semi-Soft Cheese Share by Province", pad=15)
ax.set_xlabel("Firm / Semi-Soft Cheese Share (%)")
ax.set_ylabel("Province")

for bar in bars:
    width = bar.get_width()
    ax.text(
        width + 1,
        bar.get_y() + bar.get_height() / 2,
        f"{width:.1f}%",
        va="center",
        fontweight="bold"
    )

clean_axes(ax)
ax.grid(axis="x", linestyle="--", alpha=0.25)
plt.tight_layout()
plt.show()

# ============================================================
# Final summary
# ============================================================

print("Correlation between average temperature and cheese moisture:")
print(round(merged["AvgTemperature"].corr(merged["AvgMoisture"]), 3))

print("Final Summary Table")
display(
    merged.sort_values("CheeseCount", ascending=False)
)
```

```python
# Inference for Visualization 1

Cheese Production by Province

Quebec is by far the leading Province in Canada in terms of cheese production, with Ontario and British Columbia following. This probably has to do with the fact that there are much stronger dairy industries and agricultural infrastructures in these two larger provinces, as well as established traditions of making good-quality cheese.
While some of the smaller Provinces do produce cheese, they are substantially less than what the larger Provinces do. The  smaller provinces also produce cheese based on a smaller range of cheese types, as matched with the volume of production.
```

```python
# Inference for Visualization 2

Cheese Moisture and Climate

According to the Bubble Chart, the average cheese moisture content and temperature of Canadian provinces appear to moderately relate to one another, whereby warmer provincial average temperatures result in marginally higher moisture content for cheese, whilst colder provincial average temperatures yield slightly firmer cheese that tends to have lower moisture content.
However, the relationship between the two variables does not clearly exhibit linearity as the province of Quebec produces relatively higher moisture cheeses than either British Columbia or Ontario despite possessing a colder climate than both provinces.
This indicates that other factors drive individual provincial cheese characteristics, likely including regional production preference, cheese production tradition, and consumer demand, rather than providing complete explanation via climate alone.
Additionally, the sizes of the bubbles indicating the "volume" of effects of the variable "Canadian Cheese Industry" relative to provincial averages demonstrate that provinces characterized by having a much larger cheese industry do not necessarily produce more specialty or "hard" type cheeses. The production scale and style of cheese is consequently independent factors that partially impact each other.
```

```python
# Inference for Visualization 3

Cheese Category Mix by Province

According to the percentage-based stacked bar chart, there are large differences in the number of different types of cheese across each of Canada's provinces; Quebec appears to have the most equal distribution among the cheese types, making it a very diversified province when it comes to cheese production.
The provinces of Ontario and British Columbia also have a large variety of cheese in terms of the number of different kinds produced, but smaller provinces seem to focus on producing fewer kinds of cheese. Some provinces even have large concentrations of either semi-soft or firm cheese, which may be based on the local expertise in that cheese type and/or consumer preference in the region.
Additionally, the stacked bar chart demonstrates that soft and semi-soft cheeses are produced throughout Canada, which suggests that there is a strong demand for these two types of cheese throughout the entire country regardless of what the climate is like in that province.
```

```python
# Inference for Visualization 4

Comparison of Two Temperature Sources

The historical temperature data and the weather data were closely compared, and the results show that there is a high level of agreement between the two independent climate data sets. The majority of the provinces are in close proximity to each other along a positive slope line, which provides evidence that the combined dataset used to conduct the analysis has a high level of accuracy.
This validation step provides added confidence in the analysis as it shows that there is no major bias in the measurement of climate data due to any single data set.
Some differences in the temperatures that each data set provides is to be expected since one data set represents long-term averages over time while the other represents current temperatures measured using weather stations located in various locations. However, the fact that the two data sets are well aligned demonstrates that the climate analysis results of the notebook have validity.
```

```python
# Inference for Visualization 5

Correlation Matrix: Climate and Cheese Variables

Temperature has a strong relationship with cheese moisture content; warmer climates produce cheeses with more moisture than colder; and cheeses produced in warmer climate provinces have an average moisture content of just a little more than cheeses produced elsewhere.
Two temperature variables also correlated, indicating that this supports the temperature data; therefore, thus confirming that the datasets from all of the climate data, and thus allowing the combination of temperature and cheese moisture content data to produce a valid integrated dataset.
Additionally, cheese volume produced only has a weak correlation to temperature; therefore, climate alone doesn't determine cheese production volume; rather, other elements, like the dairy infrastructure, economy investment, size of population, food culture, etc. are likely more influential than climate.
Overall, the heatmap indicates that climate may influence a few characteristics of cheese, such as moisture; however, it doesn't appear that climate is the major influence on cheese production patterns by Province.
```
